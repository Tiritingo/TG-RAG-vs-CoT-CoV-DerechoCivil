```python
# =============================================================
# CELDA 9 — Descargar archivos .rtf (corregida)
#
# Columnas reales del CSV:
#   - titulo_lista   → número de sentencia (C-665/98)
#   - detalle_url    → URL del .htm → cambiamos a .rtf
#   - fecha_providencia → fecha
#   - texto_lista    → para detectar Autos
# =============================================================

import os, re, time, random, json, requests
import pandas as pd
from tqdm.auto import tqdm

# ── Funciones ─────────────────────────────────────────────────

def clean_filename(text):
    if not text:
        text = "sin_nombre"
    text = re.sub(r'[\\/*?:"<>|]+', "_", str(text))
    text = re.sub(r"\s+", "_", text.strip())
    return text[:180]

def polite_sleep(lo=1.5, hi=3.5):
    time.sleep(random.uniform(lo, hi))

def es_auto(titulo):
    """
    Detecta Autos por el número de providencia.
    Los Autos empiezan con A. o A- (ej: A.123/24, A-045/23).
    Las sentencias empiezan con C-, T-, SU-, etc.
    """
    if not isinstance(titulo, str):
        return False
    return bool(re.match(r"^A[\.\-]", titulo.strip(), re.I))

def htm_to_rtf_url(htm_url):
    """
    Convierte la URL del .htm en URL del .rtf.
    Ejemplo:
      .../relatoria/1998/C-665-98.htm → .../relatoria/1998/C-665-98.rtf
    """
    if not htm_url or str(htm_url) == "nan":
        return None
    return str(htm_url).replace(".htm", ".rtf")

def download_rtf(url, path):
    """Descarga un .rtf en modo binario."""
    r = requests.get(url, headers=HEADERS, timeout=60)
    r.raise_for_status()
    with open(path, "wb") as f:
        f.write(r.content)

# ── Cargar índice ─────────────────────────────────────────────

idx_path = os.path.join(META_DIR, "indice_resultados.csv")
df = pd.read_csv(idx_path, encoding="utf-8-sig")
print(f"Total filas en índice  : {len(df)}")

# ── Filtrar Autos ─────────────────────────────────────────────

mask_auto = df["titulo_lista"].apply(es_auto)
df_sent   = df[~mask_auto].copy()
print(f"Autos excluidos        : {mask_auto.sum()}")

# ── Deduplicar por título (número de sentencia) ───────────────

df_sent.drop_duplicates(subset=["titulo_lista"], inplace=True)
df_sent.reset_index(drop=True, inplace=True)
print(f"Sentencias únicas      : {len(df_sent)}")

# ── Construir URL .rtf desde detalle_url ─────────────────────

df_sent["rtf_url"] = df_sent["detalle_url"].apply(htm_to_rtf_url)

print("\nEjemplos de URLs .rtf:")
print(df_sent[["titulo_lista", "fecha_providencia", "rtf_url"]].head(5).to_string(index=False))

# ── Crear carpeta rtf/ ────────────────────────────────────────

RTF_DIR = os.path.join(BASE_DIR, "rtf")
os.makedirs(RTF_DIR, exist_ok=True)
print(f"\nCarpeta destino: {RTF_DIR}")
```

```python
# ── Descarga ──────────────────────────────────────────────────

rutas, ok, errores, ya_existian, log_errores = [], 0, 0, 0, []

for _, row in tqdm(df_sent.iterrows(), total=len(df_sent), desc="Descargando .rtf"):
    titulo  = row.get("titulo_lista", "sin_nombre")
    rtf_url = row.get("rtf_url")
    slug    = clean_filename(str(titulo))
    path    = os.path.join(RTF_DIR, f"{slug}.rtf")

    if not rtf_url or str(rtf_url) == "nan":
        rutas.append(None)
        errores += 1
        log_errores.append({"titulo": titulo, "error": "sin rtf_url"})
        continue

    if os.path.exists(path):
        rutas.append(path)
        ya_existian += 1
        continue

    try:
        download_rtf(str(rtf_url), path)
        rutas.append(path)
        ok += 1
    except Exception as e:
        tqdm.write(f"  Error {titulo}: {e}")
        rutas.append(None)
        errores += 1
        log_errores.append({"titulo": titulo, "error": str(e), "url": rtf_url})

    polite_sleep(1.5, 3.5)

df_sent["rtf_local"] = rutas
```
